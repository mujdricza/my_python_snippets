# Checking values with exhaustive checks for enums

The line we use `assert_never()` should be non-reachable. If it is, it implicates a missing check like in the example below.

```python
import sys
from enum import Enum
from typing import assert_never

print(f"using Python {sys.version}")
```

```python
class Direction(Enum):
    UP = "up"
    DOWN = "down"
    NEW_NOT_CHECKED = "new"
```

```python
def choose_direction(direction: Direction) -> None:
    if direction == Direction.UP:
        print(f"Going up: {direction}")
    elif direction == Direction.DOWN:
        print(f"Going down: {direction}")
    else:
        # this line is only reachable, if direction has a value not checked in the if-elif sequence, and then it raises an AssertionError
        assert_never(direction)


```

```python
choose_direction(Direction.UP)
choose_direction(Direction.DOWN)
choose_direction(Direction.NEW_NOT_CHECKED)
```
