## General notes on magic commands (MC)

General mark for MCs: `%` before the name of the command, e.g. `%alias`.
- `%` for one-line scope
- `%%` for a whole block
- (Similar to `!` or `!!` which makes system calls.)

There are many built-in MCs, see https://ipython.readthedocs.io/en/stable/interactive/magics.html.
You can create and register your own magic commands.

The MCs are bound to the currently used IPython kernel.
- e.g. `%pip install numpy` selects the currently used environment 
- and sthus olves the misaligned Python usage issues with `!` -- with this latter, you could use an equivalent construction like this
  ```
   import sys
   !{sys.executable} -m pip install numpy
   ```

## Script usage

- `%run`: runs whole script, main-level will be executed automatically
- `%load`: loads script, script elements or lines into the code cell of the notebook
  - after running the cell, the loading command will be put as comment

````python
%run external_script.py
%run external_script.py Adam
````

````python
print("Current folder's content:")
%ls
print("Content of the 'subfolder' folder:")
%ls subfolder
print("Running script in subfolder:")
%run ./subfolder/external_script_2.py Adam
````

````python
%load external_script
"""
Small script for demonstrating script usage from within Jupyter notebook.
"""
import sys

def greeting(name: str) -> None:
    print(f"Hello, {name}!")

if __name__ == "__main__":
    name = "Anonymous"
    if len(sys.argv) > 1:
        name = sys.argv[1]
    greeting(name)
````

````python
# %load -s leaving ./subfolder/external_script_2
def leaving(name: str) -> None:
    print(f"Goodbye, {name}!")
````

## Timing
- `%timeit` for one-line code timing (with x loops)
- `%% timeit`  for timing a block (with x loops)
- `%run -t -myscript`: printing the output of the `time` command at the end of the execution -- additional `-N<N>` option loops N times

````python
r = 100

%timeit L = [n**2 for n in range(r)]
````

````python
%%timeit
L = []
for n in range(r):
    L.append(n**2)
````

````python
%run -t -N10 external_script Amadeus
````

## Package installation

- `%pip` run the pip package manager within the current kernel
- further, e.g. installing with conda: `%conda`

````python
# update pip
import sys
!{sys.executable} -m pip install --upgrade pip
````
